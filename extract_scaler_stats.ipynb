```python
import joblib
import numpy as np
import pandas as pd

# load الـ scaler
scaler = joblib.load("../model/scaler.pkl")

# load الـ feature names
feature_names = pd.read_csv("../model/feature_names.csv")["0"].tolist()

# استخراج القيم
medians   = scaler.center_
iqr_scale = scaler.scale_
iqr_upper = medians + iqr_scale

# تأكد العدد
print(f"Features in CSV:    {len(feature_names)}")
print(f"Features in scaler: {len(medians)}")
print()

# عرض الكل
print(f"{'#':<5} {'Feature':<40} {'Median':>12} {'IQR_Upper':>12}")
print("-" * 72)
for i, name in enumerate(feature_names):
    print(f"{i:<5} {name:<40} {medians[i]:>12.4f} {iqr_upper[i]:>12.4f}")

# حفظ
np.save("../model/training_medians.npy", medians)
np.save("../model/training_iqr_upper.npy", iqr_upper)

print()
print("✓ Saved: training_medians.npy")
print("✓ Saved: training_iqr_upper.npy")
```
